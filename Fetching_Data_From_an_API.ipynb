Q4. Are there duplicate movie records? If yes, identify and remove them.

```python
import pandas as pd

df = pd.read_csv("Tmdb_top_Listed_movies.csv")

# Check for duplicate records
duplicates = df.duplicated().sum()

print("Number of duplicate records:", duplicates)

# Display duplicate records
if duplicates > 0:
    print("\nDuplicate records:")
    print(df[df.duplicated()])

    # Remove duplicates
    df = df.drop_duplicates()

    print("\nDuplicates removed.")
    print("Number of records after removing duplicates:", len(df))
else:
    print("No duplicate records found.")
```

Q15. How many movies were released in each year? Show the top 10 years.

```python
import pandas as pd

df = pd.read_csv("Tmdb_top_Listed_movies.csv")

df['release_year'] = pd.to_datetime(df['release_date'], errors='coerce').dt.year

df.dropna(subset=['release_year'], inplace=True)

df['release_year'] = df['release_year'].astype(int)

top_10_years = df["release_year"].value_counts().head(10)

print("Top 10 years with the most movie releases:")
print(top_10_years)
```

Q16. Calculate the average IMDb rating for each release year.

```python
import pandas as pd

df = pd.read_csv("Tmdb_top_Listed_movies.csv")

df["release_year"] = pd.to_datetime(df["release_date"]).dt.year

average_rating = df.groupby("release_year")["vote_average"].mean()

print(average_rating)
```

Q17. Find the relationship between the number of votes and IMDb rating. Create a scatter plot.

Q21. Find the average gross revenue for movies grouped by release year.

Q22. Compare the average IMDb rating of movies with more than 100,000 votes against movies with
fewer than or equal to 100,000 votes.

```python
import pandas as pd

df = pd.read_csv("Tmdb_top_Listed_movies.csv")

more_than_100k = df[df["vote_count"] > 100000]["vote_average"].mean()

less_or_equal_100k = df[df["vote_count"] <= 100000]["vote_average"].mean()

print("Average rating for movies with more than 100,000 votes:", round(more_than_100k, 2))
print("Average rating for movies with 100,000 or fewer votes:", round(less_or_equal_100k, 2))
```
